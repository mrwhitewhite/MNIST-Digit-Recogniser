```python
import pickle
import torch
import torch.nn.functional as F
from object.model import Model as Model
from object.Data import Data as Data
import os
seed = 123
```

```python
# load data
train_path = os.path.join(os.path.abspath(os.path.join("", "..", "..")), "data", "train.csv")
test_path = os.path.join(os.path.abspath(os.path.join("", "..", "..")), "data", "test.csv")
submission_path = os.path.join(os.path.abspath(os.path.join("", "..", "..")), "data", "sample_submission.csv")


d = Data(
    train_path,
    test_path,
    submission_path,
    seed=seed,
    cv1_size=0.1,
    cv2_size=0.1,
)
```

```python
# load model
m = pickle.load(open(os.path.join(os.path.abspath(os.path.join('', "..")), "model", "model.pkl"), "rb"))
```

```python
x_test = d.test.view(-1,28,28)

with torch.no_grad():
    softmax = F.softmax(m(x_test), dim=1)
    y_test = torch.argmax(softmax, dim=1)

d.Submission['Label'] = y_test
```

```python
d.Submission.to_csv(os.path.join(os.path.abspath(os.path.join('', "..","..")), "submission", "submission.csv"), index=False)
```
